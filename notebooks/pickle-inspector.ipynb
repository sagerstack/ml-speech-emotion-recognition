```python
import pickle
from pathlib import Path

MODEL_PATH = Path('../notebooks/models/model_v1.pkl')
if not MODEL_PATH.exists():
    raise FileNotFoundError(f"Model file not found: {MODEL_PATH}")

with MODEL_PATH.open('rb') as f:
    model = pickle.load(f)

print('Loaded object type:', type(model))
print('Origin module:', getattr(model, '__module__', 'unknown'))

if hasattr(model, 'get_params'):
    print('
Model parameters:')
    for key, value in model.get_params().items():
        print(f"  {key}: {value}")

if hasattr(model, 'n_features_in_'):
    print(f"
Expected number of features: {model.n_features_in_}")
```

```python
# Inspect expected features / preprocessing steps
try:
    feature_names = getattr(model, 'feature_names_in_', None)
    if feature_names is not None:
        print('Feature names (in order):')
        for name in feature_names:
            print(' -', name)
    else:
        print('Model has no feature_names_in_ attribute.')

    if hasattr(model, 'n_features_in_'):
        print('Number of input features:', model.n_features_in_)

    if hasattr(model, 'get_feature_names_out'):
        print('get_feature_names_out:', model.get_feature_names_out())
except NameError:
    print('Run the previous cell first to load `model`.')
```
